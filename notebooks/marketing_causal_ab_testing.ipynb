### Data Cleaning and Merging Strategy

To create a comprehensive master DataFrame, we will perform the following steps:

1.  **Handle Missing Values:** Address critical missing values, especially in columns that will be used for merging or core analysis. For some columns, dropping rows or filling with a placeholder might be appropriate, depending on the context.
2.  **Convert Data Types:** Ensure all date and timestamp columns are converted to datetime objects to facilitate time-series analysis.
3.  **Merge DataFrames:** Systematically merge the DataFrames based on common keys to create a unified view of orders, customers, products, sellers, payments, and reviews. A logical merging order helps manage complexity:
    *   Start with `orders` and `order_items`.
    *   Add `products` and `product_category_name_translation`.
    *   Integrate `customers`.
    *   Incorporate `sellers`.
    *   Append `order_payments`.
    *   Add `order_reviews`.

```python
# Start with the orders DataFrame
master_df = dataframes['olist_orders_dataset'].copy()

# 1. Merge with order_items_dataset
master_df = pd.merge(master_df, dataframes['olist_order_items_dataset'], on='order_id', how='left')

# 2. Merge with products_dataset
master_df = pd.merge(master_df, dataframes['olist_products_dataset'], on='product_id', how='left')

# 3. Merge with product_category_name_translation
master_df = pd.merge(master_df, dataframes['product_category_name_translation'], on='product_category_name', how='left')

# 4. Merge with customers_dataset
master_df = pd.merge(master_df, dataframes['olist_customers_dataset'], on='customer_id', how='left')

# 5. Merge with sellers_dataset
master_df = pd.merge(master_df, dataframes['olist_sellers_dataset'], on='seller_id', how='left')

# 6. Merge with order_payments_dataset
# An order can have multiple payment types, so we'll group payments before merging
order_payments_agg = dataframes['olist_order_payments_dataset'].groupby('order_id').agg(
    payment_sequential=('payment_sequential', 'max'),
    payment_type=('payment_type', lambda x: ', '.join(x.unique())),
    payment_installments=('payment_installments', 'sum'),
    payment_value=('payment_value', 'sum')
).reset_index()
master_df = pd.merge(master_df, order_payments_agg, on='order_id', how='left')

# 7. Merge with order_reviews_dataset
# An order can have multiple reviews, so we'll group reviews before merging
order_reviews_agg = dataframes['olist_order_reviews_dataset'].groupby('order_id').agg(
    review_score=('review_score', 'mean'),
    review_comment_title=('review_comment_title', lambda x: ' '.join(x.dropna().unique()) if x.dropna().any() else None),
    review_comment_message=('review_comment_message', lambda x: ' '.join(x.dropna().unique()) if x.dropna().any() else None)
).reset_index()
master_df = pd.merge(master_df, order_reviews_agg, on='order_id', how='left')

print(f'Master DataFrame created with {master_df.shape[0]} rows and {master_df.shape[1]} columns.')
display(master_df.head())
```

## Step 3: Defining the Business Problem and Hypothesis

### 3.1 Create 'Promotional Incentive' (Voucher Usage) Variable

We define `used_voucher` to identify orders where the customer redeemed a promotional voucher/coupon as part of payment (`payment_type == 'voucher'` on at least one payment component for that order). This is a direct, literal signal of promotional-incentive exposure — unlike inferring a discount from a price/payment mismatch, which turned out to flag only 73 of 99,441 orders (too sparse for reliable causal estimates). Voucher usage covers 3,866 orders (3.9%), giving us the statistical power needed for credible matching and sensitivity analysis.

```python
# Determine, per order, whether any payment component was a voucher redemption
order_payment_types = dataframes['olist_order_payments_dataset'].groupby('order_id')['payment_type'].apply(lambda x: set(x)).reset_index()
order_payment_types['used_voucher'] = order_payment_types['payment_type'].apply(lambda s: int('voucher' in s))

# Merge the treatment flag back into master_df
master_df = pd.merge(master_df, order_payment_types[['order_id', 'used_voucher']], on='order_id', how='left')
master_df['used_voucher'] = master_df['used_voucher'].fillna(0).astype(int)

n_voucher_orders = master_df.drop_duplicates(subset='order_id')['used_voucher'].sum()
n_orders_total = master_df['order_id'].nunique()
print(f"Orders that redeemed a voucher: {n_voucher_orders:,} of {n_orders_total:,} ({n_voucher_orders / n_orders_total:.2%})")
print("First 5 rows of master_df with new 'used_voucher' column:")
display(master_df[['order_id', 'payment_type', 'used_voucher']].head())
```

### 3.2 Define the Conversion Outcome: Repeat Purchase (Retention)

Every row in this dataset is already a *completed* order, so there is no browse-vs-buy funnel to measure "conversion" against. Instead we use **repeat purchase** as the conversion/retention outcome: for each order, did that customer (`customer_unique_id`) go on to place another order afterward? This is the standard proxy for marketing-driven conversion in a post-purchase transactional dataset, and it's the outcome we'll carry through the power analysis and causal inference sections below.

```python
# Build a de-duplicated order-level table with customer identity and purchase timestamp
order_customer = master_df[['order_id', 'customer_unique_id', 'order_purchase_timestamp']].drop_duplicates(subset='order_id').copy()
order_customer = order_customer.sort_values(['customer_unique_id', 'order_purchase_timestamp'])

# An order counts as a "conversion" (repeat_purchase = 1) if the same customer placed
# at least one more order after it
order_customer['order_rank'] = order_customer.groupby('customer_unique_id').cumcount()
orders_per_customer = order_customer.groupby('customer_unique_id')['order_id'].transform('count')
order_customer['repeat_purchase'] = (order_customer['order_rank'] < (orders_per_customer - 1)).astype(int)

baseline_repeat_rate = order_customer['repeat_purchase'].mean()
print(f"Unique customers: {order_customer['customer_unique_id'].nunique():,}")
print(f"Orders followed by a repeat purchase: {order_customer['repeat_purchase'].sum():,} of {len(order_customer):,} ({baseline_repeat_rate:.2%})")
display(order_customer.head())
```

### 3.3 A/B Test Power Analysis

Before (hypothetically) running a prospective A/B test on promotional incentives, we size the experiment: how many customers per arm are needed to detect a **3 percentage-point minimum detectable effect (MDE)** on the repeat-purchase rate, at **95% confidence (α = 0.05, two-sided)** and **80% power**, using the *actual observed* baseline repeat-purchase rate from historical data.

```python
from statsmodels.stats.power import NormalIndPower
from statsmodels.stats.proportion import proportion_effectsize

MDE_ABSOLUTE = 0.03  # 3 percentage-point minimum detectable effect
ALPHA = 0.05
POWER = 0.80

p1 = baseline_repeat_rate
p2 = baseline_repeat_rate + MDE_ABSOLUTE

effect_size = proportion_effectsize(p2, p1)
power_analysis = NormalIndPower()
n_per_group = power_analysis.solve_power(effect_size=effect_size, alpha=ALPHA, power=POWER, ratio=1, alternative='two-sided')

print(f"Baseline repeat-purchase rate:            {p1:.2%}")
print(f"Target repeat-purchase rate (+{MDE_ABSOLUTE:.0%} MDE):    {p2:.2%}")
print(f"Effect size (Cohen's h):                  {effect_size:.4f}")
print(f"Required sample size per arm:              {n_per_group:,.0f}")
print(f"Required total sample size:                {n_per_group * 2:,.0f}")
print(f"Orders available in historical dataset:    {len(order_customer):,}")
print(f"Feasible with historical volume: {'Yes' if len(order_customer) >= n_per_group * 2 else 'No'}")
```

### 3.4 Identify Confounding Variables

Now that treatment (`used_voucher`) and the outcome (`repeat_purchase`, with `review_score` as a secondary check) are defined, we identify confounders: variables that plausibly influence both whether a customer used a voucher and whether they came back to buy again. Note `payment_type` is **excluded** from the confounder list here, since it's the source column the treatment itself is derived from — including it would be circular (data leakage), not genuine confounding control.

```python
confounding_variables = [
    'product_category_name_english',
    'customer_state',
    'seller_state',
    'price', # Original item price as a baseline
    'freight_value', # Original item freight as a baseline
    'payment_installments',
    'order_item_id' # Number of items in order
    # payment_type intentionally excluded: used_voucher is derived from it (would leak treatment)
]

# Bring repeat_purchase onto the item-level master_df for descriptive comparisons
master_df = master_df.merge(order_customer[['order_id', 'repeat_purchase']], on='order_id', how='left')

# Display summary statistics for repeat-purchase rate and review score by voucher usage
print("\n--- Repeat-Purchase Rate vs. Voucher Usage ---\n")
display(master_df.groupby('used_voucher')['repeat_purchase'].agg(['mean', 'count']))

print("\n--- Review Score vs. Voucher Usage ---\n")
display(master_df.groupby('used_voucher')['review_score'].agg(['mean', 'median', 'std', 'count']))

# Check distribution of a key confounder (product category) by voucher usage
print("\n--- Mean Review Score by Product Category ---\n")
display(master_df.groupby('product_category_name_english')['review_score'].mean().sort_values(ascending=False).head(10))
```

### 4.2 Estimate Propensity Scores and Perform Matching

We use logistic regression on the confounders above to estimate each order's propensity score (probability of voucher usage), then match voucher-using orders to their nearest non-voucher counterpart on that score (1:1 nearest-neighbor matching with bias adjustment). This estimates the Average Treatment Effect (ATE/ATT) of `used_voucher` on our primary outcome (`repeat_purchase`) and, separately, on `review_score` as a secondary check.

```python
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.neighbors import NearestNeighbors
from causalinference import CausalModel
import numpy as np

# 1. Aggregate to order level, bringing in repeat_purchase (primary outcome) and review_score (secondary)
# payment_type is intentionally NOT aggregated here as a confounder: used_voucher is derived from it
order_level_df = master_df.groupby('order_id').agg(
    used_voucher=('used_voucher', 'first'),
    review_score=('review_score', 'mean'),
    product_category_name_english=('product_category_name_english', lambda x: x.mode()[0] if not x.mode().empty else 'unknown'),
    customer_state=('customer_state', 'first'),
    seller_state=('seller_state', 'first'),
    price=('price', 'sum'),
    freight_value=('freight_value', 'sum'),
    payment_installments=('payment_installments', 'mean'),
    num_items=('order_item_id', 'count')
).reset_index()

order_level_df = order_level_df.merge(order_customer[['order_id', 'repeat_purchase']], on='order_id', how='left')
order_level_df = order_level_df.dropna(subset=['review_score', 'used_voucher', 'repeat_purchase'])

# 2. Select and prepare confounding variables (payment_type excluded: would leak treatment)
confounding_variables_order_level = [
    'product_category_name_english',
    'customer_state',
    'seller_state',
    'price',
    'freight_value',
    'payment_installments',
    'num_items'
]
confounding_variables_order_level_present = [col for col in confounding_variables_order_level if col in order_level_df.columns]

X_temp_ol = order_level_df[confounding_variables_order_level_present].copy()

# Reduce cardinality for high-cardinality categorical features
top_n_product_categories = 5
top_n_customer_states = 3
top_n_seller_states = 3

top_product_categories = X_temp_ol['product_category_name_english'].value_counts().index[:top_n_product_categories].tolist()
X_temp_ol['product_category_name_english'] = X_temp_ol['product_category_name_english'].apply(lambda x: x if x in top_product_categories else 'Other')

top_customer_states = X_temp_ol['customer_state'].value_counts().index[:top_n_customer_states].tolist()
X_temp_ol['customer_state'] = X_temp_ol['customer_state'].apply(lambda x: x if x in top_customer_states else 'Other')

top_seller_states = X_temp_ol['seller_state'].value_counts().index[:top_n_seller_states].tolist()
X_temp_ol['seller_state'] = X_temp_ol['seller_state'].apply(lambda x: x if x in top_seller_states else 'Other')

# 3. Preprocessing
categorical_features_ol = X_temp_ol.select_dtypes(include=['object']).columns
numerical_features_ol = X_temp_ol.select_dtypes(include=['number']).columns

numerical_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='mean')),
    ('scaler', StandardScaler())
])

categorical_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='constant', fill_value='missing')),
    ('onehot', OneHotEncoder(handle_unknown='ignore'))
])

preprocessor_ol = ColumnTransformer(
    transformers=[
        ('num', numerical_transformer, numerical_features_ol),
        ('cat', categorical_transformer, categorical_features_ol)
    ])

X_ol = preprocessor_ol.fit_transform(X_temp_ol)
if hasattr(X_ol, 'toarray'):
    X_ol = X_ol.toarray()

D_ol = order_level_df['used_voucher'].values.astype(int)
Y_repeat = order_level_df['repeat_purchase'].values.astype(float)
Y_review = order_level_df['review_score'].values.astype(float)

# 4. Estimate propensity scores via logistic regression on the full covariate matrix
lr = LogisticRegression(solver='liblinear', random_state=42)
lr.fit(X_ol, D_ol)
propensity_scores = lr.predict_proba(X_ol)[:, 1].reshape(-1, 1)

# 5. Primary causal estimate: effect of used_voucher on repeat_purchase (retention/conversion)
causal_repeat = CausalModel(Y_repeat, D_ol, propensity_scores)
print("--- Covariate Balance BEFORE Matching (normalized differences) ---\n")
print(causal_repeat.summary_stats)

causal_repeat.est_via_matching(matches=1, bias_adj=True)
print("\n--- Causal Inference Results: used_voucher -> repeat_purchase (PRIMARY) ---\n")
print(causal_repeat.estimates)

# 6. Secondary robustness estimate: effect of used_voucher on review_score (satisfaction)
causal_review = CausalModel(Y_review, D_ol, propensity_scores)
causal_review.est_via_matching(matches=1, bias_adj=True)
print("\n--- Causal Inference Results: used_voucher -> review_score (SECONDARY) ---\n")
print(causal_review.estimates)
```

### 4.3 Sensitivity Analysis: Rosenbaum Bounds

PSM removes bias from *observed* confounders, but says nothing about **unobserved** ones (e.g. a customer's latent price-sensitivity that could drive both `had_discount` and `repeat_purchase`). Rosenbaum bounds ask: how strong would a hidden confounder have to be (in terms of odds-ratio Γ) to overturn our conclusion that the treatment effect is significantly different from zero? We construct 1:1 nearest-neighbor matched pairs on the propensity score, then re-run the Wilcoxon signed-rank test under increasing assumed levels of hidden bias Γ.

```python
from scipy.stats import wilcoxon

# Build 1:1 nearest-neighbor matched pairs on the propensity score (treated -> nearest control)
ps = propensity_scores.ravel()
treated_idx = np.where(D_ol == 1)[0]
control_idx = np.where(D_ol == 0)[0]

nn = NearestNeighbors(n_neighbors=1).fit(ps[control_idx].reshape(-1, 1))
_, nn_pos = nn.kneighbors(ps[treated_idx].reshape(-1, 1))
matched_control_idx = control_idx[nn_pos.ravel()]

pair_diffs = Y_repeat[treated_idx] - Y_repeat[matched_control_idx]
nonzero_diffs = pair_diffs[pair_diffs != 0]

print(f"Matched pairs: {len(pair_diffs):,} (non-tied pairs used in test: {len(nonzero_diffs):,})")

def rosenbaum_bounds_pvalue(diffs, gamma):
    """Wilcoxon signed-rank test upper-bound p-value under hidden-bias odds ratio gamma."""
    p_plus = gamma / (1 + gamma)
    n = len(diffs)
    ranks = pd.Series(np.abs(diffs)).rank().values
    signs = np.where(diffs > 0, 1, 0)
    W = np.sum(ranks * signs)
    mean_W = p_plus * np.sum(ranks)
    var_W = p_plus * (1 - p_plus) * np.sum(ranks ** 2)
    z = (W - mean_W) / np.sqrt(var_W)
    from scipy.stats import norm
    return 1 - norm.cdf(z)

print("\n--- Rosenbaum Bounds: upper-bound p-value as hidden bias (Gamma) increases ---\n")
for gamma in [1.0, 1.1, 1.25, 1.5, 1.75, 2.0]:
    p_val = rosenbaum_bounds_pvalue(nonzero_diffs, gamma)
    print(f"Gamma = {gamma:.2f}:  upper-bound p-value = {p_val:.4f}")
```

### 4.4 Interpretation of Results

**Primary finding — voucher usage → repeat purchase:**
After matching on product category, customer/seller state, price, freight, payment installments, and order size, voucher-using orders show an **ATT of +1.0 percentage point** on the repeat-purchase rate (95% CI: -0.1pp to +2.2pp, p = 0.068) against a **3.36% baseline** — a relative lift of roughly **+30%**, but only *marginally* significant at conventional 95% confidence (it clears the 90% threshold, not 95%). The population-average effect (ATE = +0.5pp, p = 0.393) is not statistically distinguishable from zero. This is meaningfully weaker evidence than a clean, confidently-isolated causal lift.

**Secondary finding — voucher usage → review score:**
The raw population effect (ATE = -0.093, p = 0.014) is significantly *negative*, but the matched-pair effect (ATT = -0.024, p = 0.556) is not significant. Read together, this says: voucher users have lower satisfaction on average, but most of that gap is explained by *who* uses vouchers (the confounders), not by the voucher itself. Voucher usage shows no credible positive or negative causal effect on satisfaction once we control for observed differences.

**Robustness — Rosenbaum bounds:**
The repeat-purchase result is fragile to hidden bias. At Gamma = 1.00 (no hidden bias) the upper-bound p-value is 0.014 (significant); by **Gamma ≈ 1.10** — an unobserved confounder that shifts the odds of voucher redemption by just 10% — the bound already crosses p = 0.05, and by Gamma = 1.25 the result is fully explainable by hidden bias (p = 0.36). In practice this means an unmeasured factor as mild as "slightly more price-conscious shoppers self-select into both voucher use and repeat purchase" could account for the entire estimated effect.

**Power analysis check:**
This is *not* an underpowered study — 3,866 treated orders vs. the ~1,576 total required for an 80%-powered 3pp MDE test comfortably clears the bar. The marginal significance reflects a genuinely small, fragile effect, not insufficient sample size.

**Honest bottom line vs. the original hypothesis:** this analysis does **not** reproduce a clean "12% lift, causally isolated" result on Olist's real data — that specific number describes a different (undisclosed) dataset/engagement. What the historical Olist data actually supports is: *suggestive but not robust* evidence (p ≈ 0.07, collapses under mild hidden-bias sensitivity) that voucher redemption is associated with a ~30% relative lift in repeat-purchase rate, with no credible effect on satisfaction. The value of this notebook as a portfolio piece is in demonstrating the full methodology — proxy construction, power analysis, propensity-score matching with covariate balance diagnostics, and Rosenbaum-bounds sensitivity testing — rigorously and transparently, including reporting a result that doesn't fully hold up under scrutiny, which is itself the point of a sensitivity analysis.
